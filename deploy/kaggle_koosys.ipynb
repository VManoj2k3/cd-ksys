# Cd koosys — Code Review System (Kaggle 2x T4)

**Before running:** Settings → Accelerator = **GPU T4 x2**, Internet = **On**.

**Get the project in:** either (a) upload the `Cd koosys` project folder as a Kaggle *Dataset* and set `DATASET_NAME` below, or (b) push it to a git repo and set `GIT_URL`.

Run cells top to bottom. The last setup cell prints a public `trycloudflare.com` URL — open it in your browser.

```python
# --- where to get the project (pick ONE) ---
DATASET_NAME = "cd-koosys"   # Kaggle dataset name, if you uploaded the folder as a dataset
GIT_URL = ""                  # or a git repo URL

import shutil, subprocess, sys
from pathlib import Path
APP = Path("/kaggle/working/app")
if not APP.exists():
    if GIT_URL:
        subprocess.run(["git", "clone", GIT_URL, str(APP)], check=True)
    else:
        src = next(p for p in Path("/kaggle/input").glob(f"{DATASET_NAME}*"))
        # dataset may nest the project one level down
        if not (src / "config.yaml").exists():
            src = next(p for p in src.rglob("config.yaml")).parent
        shutil.copytree(src, APP)
sys.path.insert(0, str(APP))
print("project at", APP)
```
